```python
r1 = [1, 1, -1]
r2 = [1, 1, 1]

R1, R2 = 0.0,0.0

```

# No Bootstrapping

```python
for i in range(len(r1)-1,0,-1):  
    R1 = r1[i] + 0.99*R1
    
for i in range(len(r2)-1,0,-1):  
    R2 = r2[i] + 0.99*R2
```

```python
print("No bootstrapping") 
print(R1,R2)
```

# With bootstrapping

```python
R1,R2 = 1.0,1.0
```

```python
for i in range(len(r1)-1,0,-1):  R1 = r1[i] + 0.99*R1
for i in range(len(r2)-1,0,-1):  R2 = r2[i] + 0.99*R2
```

```python
print("With bootstrapping") 
print(R1,R2)
```
